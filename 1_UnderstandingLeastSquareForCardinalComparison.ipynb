```python
import numpy as np
import itertools
import pandas as pd

def generate_pcg_data(K, edge_prob=0.5, min_comparisons=1, max_comparisons=5, sigma=1.0, seed=None):
    """
    Generate Pairwise Comparison Graph (PCG) data

    Parameters:
        K (int): number of items
        edge_prob (float): probability that an edge exists between i and j
        min_comparisons (int): minimum nij
        max_comparisons (int): maximum nij
        sigma (float): noise std deviation
        seed (int): random seed

    Returns:
        mu (np.array): true scores (size K)
        edges (list): list of edges (i, j)
        data (dict): {(i,j): [Y_ij1, ..., Y_ijnij]}
    """
    if seed is not None:
        np.random.seed(seed)

    # Step 1: Generate true merits (μ)
    mu = np.random.randn(K)

    edges = []
    data = {}
    data_matrix = np.full((K, K), None, dtype=object)
    n_ij = np.zeros((K,K))

    # Step 2: Generate graph + observations
    for i, j in itertools.combinations(range(K), 2):
      # creates 01 02 03 04 12 13 14 23 24 34  k = 5

        if np.random.rand() < edge_prob:
            edges.append((i, j))

            nij = np.random.randint(min_comparisons, max_comparisons + 1)
            # random int btw 1, 6

            # Generate comparisons and noise based on mean and sigma
            noise = np.random.normal(0, sigma, size=nij)

            Y_ij = (mu[i] - mu[j]) + noise

            data[(i, j)] = Y_ij
            data[(j, i)] = -Y_ij
            data_matrix[i][j] = Y_ij
            data_matrix[j][i] = -Y_ij


            n_ij[i][j] = n_ij[j][i] = nij

        # else:
        #     data[(i,j)]= data[(j,i)] = None

    return mu, edges, data, n_ij, data_matrix


def generate_pcg_data_startstop4mu(K, edge_prob=0.5, min_comparisons=1, max_comparisons=5, sigma=1.0, seed=None, start = -2, stop = 2):
    """
    Generate Pairwise Comparison Graph (PCG) data

    Parameters:
        K (int): number of items
        edge_prob (float): probability that an edge exists between i and j
        min_comparisons (int): minimum nij
        max_comparisons (int): maximum nij
        sigma (float): noise std deviation
        seed (int): random seed

    Returns:
        mu (np.array): true scores (size K)
        edges (list): list of edges (i, j)
        data (dict): {(i,j): [Y_ij1, ..., Y_ijnij]}
    """
    if seed is not None:
        np.random.seed(seed)

    # Step 1: Generate true merits (μ)
    mu = np.linspace(start, stop, K)

    edges = []
    data = {}
    data_matrix = np.full((K, K), None, dtype=object)
    n_ij = np.zeros((K,K))

    # Step 2: Generate graph + observations
    for i, j in itertools.combinations(range(K), 2):
      # creates 01 02 03 04 12 13 14 23 24 34  k = 5

        if np.random.rand() < edge_prob:
            edges.append((i, j))

            nij = np.random.randint(min_comparisons, max_comparisons + 1)

            # random int btw 1, 6

            # Generate comparisons and noise based on mean and sigma
            noise = np.random.normal(0, sigma, size=nij)

            Y_ij = (mu[i] - mu[j]) + noise

            data[(i, j)] = Y_ij
            data[(j, i)] = -Y_ij
            data_matrix[i][j] = Y_ij
            data_matrix[j][i] = -Y_ij


            n_ij[i][j] = n_ij[j][i] = nij

        # else:
        #     data[(i,j)]= data[(j,i)] = None

    return mu, edges, data, n_ij, data_matrix



def generate_pcg_data_nij2inf(nij2inf, K, edge_prob=0.5, min_comparisons=1, max_comparisons=5, sigma=1.0, seed=None, start = -2, stop = 2):
    """
    Generate Pairwise Comparison Graph (PCG) data

    Parameters:
        K (int): number of items
        edge_prob (float): probability that an edge exists between i and j
        min_comparisons (int): minimum nij
        max_comparisons (int): maximum nij
        sigma (float): noise std deviation
        seed (int): random seed

    Returns:
        mu (np.array): true scores (size K)
        edges (list): list of edges (i, j)
        data (dict): {(i,j): [Y_ij1, ..., Y_ijnij]}
    """
    if seed is not None:
        np.random.seed(seed)

    # Step 1: Generate true merits (μ)
    mu = np.linspace(start, stop, K)

    edges = []
    data = {}
    data_matrix = np.full((K, K), None, dtype=object)
    n_ij = np.zeros((K,K))

    # Step 2: Generate graph + observations
    for i, j in itertools.combinations(range(K), 2):
      # creates 01 02 03 04 12 13 14 23 24 34  k = 5

        if np.random.rand() < edge_prob:
            edges.append((i, j))

            if np.random.rand() < 0.9:
              nij = np.random.randint(min_comparisons, max_comparisons + 1)
            else:
              nij = np.random.randint(nij2inf, nij2inf*2)
            # random int btw 1, 6 or nij grows to inf

            # Generate comparisons and noise based on mean and sigma
            noise = np.random.normal(0, sigma, size=nij)

            Y_ij = (mu[i] - mu[j]) + noise

            data[(i, j)] = Y_ij
            data[(j, i)] = -Y_ij
            data_matrix[i][j] = Y_ij
            data_matrix[j][i] = -Y_ij


            n_ij[i][j] = n_ij[j][i] = nij

        # else:
        #     data[(i,j)]= data[(j,i)] = None

    return mu, edges, data, n_ij, data_matrix

```

```python
'''

building laplacian on the concept of Nij =

j nij if i = j,
−nij if i ≠ j,

'''

def build_laplcian(n_ij):
  K = n_ij.shape[0]
  L = np.zeros((K,K))

  for i in range(K):
    for j in range(K):
      if i == j:
        L[i][j] = np.sum(n_ij[i])
      else:
        L[i][j] = -n_ij[i][j]

  return L

build_laplcian(n_ij)
```

```python
'''
Si = 􏽐
j≠i Sij
'''

def create_vector(data, K):
  vector = np.zeros(K)
  for key,value in data.items():
    # print(key)
    i = key[0]
    j = key[1]
    vector[i] += np.sum(value)
    # vector[j] -= np.sum(value)
  return vector

create_vector(data,K)
```

```python
def estimate_mu(data, K, n_ij):
  N = build_laplcian(n_ij)
  vector = create_vector(data, K)

  # Step 1: Compute pseudoinverse
  print('N', N)
  print('vector', vector)
  N_pinv = np.linalg.pinv(N)

  # Multiply the N+ with S
  raw_mu = N_pinv @ vector
  return raw_mu

estimate_mu(data, K, n_ij)
```

```python
raw_mu = estimate_mu(data, K, n_ij)
def rank_scores(scores):
  rank = [[i,scores[i]] for i in range(len(scores))]
  rank = [raw_mu[0] for raw_mu in sorted(rank, key=lambda x: -x[1])]
  return rank

```

```python
rank_calculated = rank_scores(raw_mu)
rank = rank_scores(mu)
rank_calculated, rank
```

```python
# calculating the caley distance

def caley_distance(rank, rank_calculated):

  l = len(rank)
  cycles = 0

  pos_dict = {val:idx for idx, val in enumerate(rank_calculated)}
  rank_pos = [pos_dict[val] for val in rank]

  # print(rank_pos)

  visited = [False] * l

  for i in range(l):
    if not visited[i]:
      cycles+=1
      j = i
      while not visited[j]:
        visited[j] = True
        j = rank_pos[j]



  return l - cycles, cycles

caley_distance(rank, rank_calculated)
```

```python
from scipy.stats import kendalltau
tau, _ = kendalltau(rank, rank_calculated)

# Convert correlation → distance
distance = (1 - tau) / 2

print("Kendall tau:", tau)
print("Distance:", distance)
```

Let's perform a Monte Carlo simulation to analyze the average Kendall Tau distance and Cayley distance for a fixed `K` over multiple random data generations. We will run the simulation `num_simulations` times, each with a different random seed.

```python
num_simulations = 10 # Number of Monte Carlo runs


monte_carlo_results = {
    'K' : [],
    'simulation_run': [],
    'kendall_tau': [],
    'cayley_distance': []
}

for i in range(1, num_simulations):
    # Generate a new random seed for each simulation run
    current_seed = np.random.randint(1, 10000)
    K_fixed = 5*i
    # Generate data with the current seed
    mu, edges, data, n_ij, data_matrix = generate_pcg_data_startstop4mu(
        K_fixed, edge_prob=0.7, seed=current_seed
    )

    # Estimate mu and calculate ranks
    raw_mu = estimate_mu(data, K_fixed, n_ij)
    rank_calculated = rank_scores(raw_mu)
    rank_true = rank_scores(mu)

    # Calculate metrics
    tau, _ = kendalltau(rank_true, rank_calculated)
    cayley_dist, _ = caley_distance(rank_true, rank_calculated)

    # Store results
    monte_carlo_results['K'].append(K_fixed)
    monte_carlo_results['simulation_run'].append(i + 1)
    monte_carlo_results['kendall_tau'].append(tau)
    monte_carlo_results['cayley_distance'].append(cayley_dist)

df_monte_carlo = pd.DataFrame(monte_carlo_results)

print(f"Monte Carlo simulation completed with {num_simulations} runs for K tending to infinity.")
print("Average Kendall Tau:", df_monte_carlo['kendall_tau'].mean())
print("Variance Kendall Tau:", df_monte_carlo['kendall_tau'].var())
print("Average Cayley Distance:", df_monte_carlo['cayley_distance'].mean())
print("Variance Cayley Distance:", df_monte_carlo['cayley_distance'].var())
df_monte_carlo
```

```python
import numpy as np
import itertools

def generate_binary_covariate_data(K, edge_prob=0.7, sigma=1.0, seed=42):
    np.random.seed(seed)

    mu_true = np.linspace(-2, 2, K)
    beta_true = np.array([2.0])  # effect of "being home"

    rows = []

    for i, j in itertools.combinations(range(K), 2):
        if np.random.rand() < edge_prob:
            nij = np.random.randint(1, 6)

            for _ in range(nij):
                # Randomly decide who is "home"
                if np.random.rand() < 0.5:
                    x_i = 1
                    x_j = 0
                else:
                    x_i = 0
                    x_j = 1

                x_ijk = np.array([x_i - x_j])  # will be +1 or -1

                noise = np.random.normal(0, sigma)
                y = mu_true[i] - mu_true[j] + beta_true @ x_ijk + noise

                rows.append((i, j, y, x_ijk))

    return mu_true, beta_true, rows
```

```python
def build_matrices(K, rows):
    n = len(rows)

    M = np.zeros((n, K))
    X = np.zeros((n, 1))  # only 1 binary feature
    Y = np.zeros(n)

    for idx, (i, j, y, x_ijk) in enumerate(rows):
        M[idx, i] = 1
        M[idx, j] = -1
        X[idx] = x_ijk
        Y[idx] = y

    return M, X, Y
```

```python
K = 5
mu_true, beta_true, rows = generate_binary_covariate_data(K)
M, X, Y = build_matrices(K, rows)

mu_true, beta_true, rows, M, X, Y
```

```python
# estimate mu hat and beta hat

def estimate_parameters(M, X, Y):
    H = np.hstack([M, X])

    theta_hat = np.linalg.lstsq(H, Y, rcond=None)[0]

    K = M.shape[1]

    mu_hat = theta_hat[:K]
    beta_hat = theta_hat[K:]

    # enforce sum(mu)=0
    mu_hat = mu_hat - np.mean(mu_hat)

    return mu_hat, beta_hat
```

```python
K = 100

mu_true, beta_true, rows = generate_binary_covariate_data(K)

M, X, Y = build_matrices(K, rows)
mu_hat, beta_hat = estimate_parameters(M, X, Y)

print("True mu:", mu_true)
print("Estimated mu:", mu_hat)

print("True beta:", beta_true)
print("Estimated beta:", beta_hat)

rank_true = rank_scores(mu_true)
rank_hat = rank_scores(mu_hat)

print("True rank:", rank_true)
print("Estimated rank:", rank_hat)

tau, _ = kendalltau(rank_true, rank_hat)
print("Kendall tau:", tau)

```
