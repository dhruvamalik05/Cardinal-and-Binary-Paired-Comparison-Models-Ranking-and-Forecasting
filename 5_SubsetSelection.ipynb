# **Best Subset Selection**

```python
def cardinal_prepare_ipl_data_for_year_with_covariates(merged_df, iplyear, selected_covariates_names):
  ipl_year_df = merged_df[merged_df['year'] == iplyear].copy()

  unique_teams = sorted(list(pd.concat([ipl_year_df['team1'], ipl_year_df['team2']]).unique()))
  team_to_id = {team: i for i, team in enumerate(unique_teams)}
  K_teams = len(unique_teams)

  num_matches = len(ipl_year_df)
  num_covariates = len(selected_covariates_names)

  M = np.zeros((num_matches, K_teams))
  X_covariates = np.zeros((num_matches, num_covariates))
  Y = np.zeros(num_matches)

  for idx, (index, row) in enumerate(ipl_year_df.iterrows()):
      team1_id = team_to_id[row['team1']]
      team2_id = team_to_id[row['team2']]

      M[idx, team1_id] = 1
      M[idx, team2_id] = -1

      for i, cov_name in enumerate(selected_covariates_names):
          X_covariates[idx, i] = row[cov_name]

      Y[idx] = row['win_by_runs']

  return M, X_covariates, Y, team_to_id, K_teams

def iterate_covariate_combinations(merged_df_covariates, iplyear):
    all_covariates = [
        'home_ground_advantage',
        'days_since_last_match_diff',
        'previous_match_advantage',
        'toss_winner_encoded',
        'rank_difference_prev_year'
    ]

    results = []

    # Iterate through all 2^N combinations of covariates
    for i in range(2**len(all_covariates)):
        selected_covariates_mask = [
            bool((i >> j) & 1) for j in range(len(all_covariates))
        ]
        current_covariates = [
            all_covariates[j] for j, include in enumerate(selected_covariates_mask) if include
        ]


        if not current_covariates:
            # Handle the case with no covariates explicitly
            M, X_empty, Y, team_to_id, K = cardinal_prepare_ipl_data_for_year_with_covariates(merged_df_covariates, iplyear, [] )
            mu_hat, _, _ = estimate_parameters(M, np.zeros((M.shape[0], 0)), Y)
        else:
            M, X, Y, team_to_id, K = cardinal_prepare_ipl_data_for_year_with_covariates(merged_df_covariates, iplyear, current_covariates)
            mu_hat, _, _ = estimate_parameters(M, X, Y)
        mu_hat = mu_hat - np.mean(mu_hat)
        rank_hat = rank_scores(mu_hat)

        # kend_yr - kendalltau (true_rank_yr, cal_rank_yr),
        # true_rank_yr - true rank from IPL website
        # cal_rank_yr - calculated rank of the year nocovariate
        # rank_hat - calculated rank of the year with the covariate subset

        kend_yr, true_rank_yr, cal_rank_yr = cal_rank(iplyear)
        kendall_tau_val, _ = kendalltau(cal_rank_yr, rank_hat)

        results.append({
            'year': iplyear,
            'covariates': current_covariates if current_covariates else 'None',
            'cardinal nocovariate and covariate kendall_tau': kendall_tau_val
        })

    return pd.DataFrame(results)


all_years_results = []
for year in range(2008, 2020):
    print(f"Running covariate combinations analysis for year {year}...")
    year_results_df = iterate_covariate_combinations(merged_df_covariates, year)
    all_years_results.append(year_results_df)

combined_results_df = pd.concat(all_years_results, ignore_index=True)
combined_results_df['covariates_str'] = combined_results_df['covariates'].apply(lambda x: str(x))
pivoted_kendall_tau = combined_results_df.pivot_table(
    index='covariates_str',
    columns='year',
    values='cardinal nocovariate and covariate kendall_tau'
)

print("Pivot table with years as columns and covariate combinations as rows (Kendall Tau values):")
display(pivoted_kendall_tau)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# =========================================================
# BEST MODEL SELECTION USING:
# R², Adjusted R², AIC, BIC
# =========================================================


def evaluate_subsets_information_criteria_statsmodels(merged_df_covariates,iplyear):

    all_covariates = [
        'home_ground_advantage',
        'days_since_last_match_diff',
        'previous_match_advantage',
        'toss_winner_encoded',
        'rank_difference_prev_year'
    ]

    results = []

    # -----------------------------------------------------
    # ITERATE OVER ALL 2^5 SUBSETS
    # -----------------------------------------------------

    for i in range(2 ** len(all_covariates)):
        selected_covariates_mask = [bool((i >> j) & 1)for j in range(len(all_covariates))]
        current_covariates = [all_covariates[j] for j, include in enumerate(selected_covariates_mask) if include]

        M, X, Y, team_to_id, K = cardinal_prepare_ipl_data_for_year_with_covariates(merged_df_covariates,iplyear,current_covariates)

        if len(current_covariates) == 0:
            H = M
        else:
            H = np.hstack([M, X])

        # This model works in the same fashion as np.linalg.lstsq
        model = sm.OLS(Y, H).fit()
        theta_hat = model.params
        mu_hat = theta_hat[:K]
        mu_hat = mu_hat - np.mean(mu_hat)

        results.append({

            "year": iplyear,

            "covariates":
                ", ".join(current_covariates)
                if current_covariates else "None",

            "r2":
                model.rsquared,

            "adj_r2":
                model.rsquared_adj,

            "aic":
                model.aic,

            "bic":
                model.bic
        })

    return pd.DataFrame(results)

def best_models_by_metric(results_df):

    # -----------------------------------------------------
    # MAX R²
    # -----------------------------------------------------

    max_r2 = results_df["r2"].max()
    best_r2_rows = results_df[np.isclose(results_df["r2"], max_r2)]

    # -----------------------------------------------------
    # MAX ADJUSTED R²
    # -----------------------------------------------------

    max_adj_r2 = results_df["adj_r2"].max()
    best_adj_r2_rows = results_df[np.isclose(results_df["adj_r2"], max_adj_r2)]

    # -----------------------------------------------------
    # MIN AIC
    # -----------------------------------------------------

    min_aic = results_df["aic"].min()
    best_aic_rows = results_df[np.isclose(results_df["aic"], min_aic)]

    # -----------------------------------------------------
    # MIN BIC
    # -----------------------------------------------------

    min_bic = results_df["bic"].min()
    best_bic_rows = results_df[np.isclose(results_df["bic"], min_bic)]

    return {

        "best_r2":
            best_r2_rows,

        "best_adj_r2":
            best_adj_r2_rows,

        "best_aic":
            best_aic_rows,

        "best_bic":
            best_bic_rows
    }


# =========================================================
# RUN ACROSS ALL IPL YEARS
# =========================================================

all_years_model_selection = []

for year in range(2008, 2020):

    print(f"Running subset evaluation for {year}...")


    results_df = evaluate_subsets_information_criteria_statsmodels(merged_df_covariates,year)
    best_models = best_models_by_metric(results_df)
    all_years_model_selection.append({

        "year": year,

        "best_r2_covariates":
            best_models["best_r2"]["covariates"].tolist(),

        "best_r2":
            best_models["best_r2"]["r2"].iloc[0],

        "best_adj_r2_covariates":
            best_models["best_adj_r2"]["covariates"].tolist(),

        "best_adj_r2":
            best_models["best_adj_r2"]["adj_r2"].iloc[0],

        "best_aic_covariates":
            best_models["best_aic"]["covariates"].tolist(),

        "best_aic":
            best_models["best_aic"]["aic"].iloc[0],

        "best_bic_covariates":
            best_models["best_bic"]["covariates"].tolist(),

        "best_bic":
            best_models["best_bic"]["bic"].iloc[0]
    })


best_subset_summary_df = pd.DataFrame(all_years_model_selection)
display(best_subset_summary_df)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import kendalltau

# PREPARE BRADLEY-TERRY DATA WITH SELECTED COVARIATES


# =========================================================
# EXHAUSTIVE SUBSET SELECTION FOR BRADLEY-TERRY
# =========================================================

def evaluate_bt_subsets(df, year):
    all_covariates = [
        'home_ground_advantage',
        'days_since_last_match_diff',
        'previous_match_advantage',
        'toss_winner_encoded',
        'rank_difference_prev_year'
    ]

    results = []

    for i in range(2 ** len(all_covariates)):
        selected_covariates_mask = [bool((i >> j) & 1) for j in range(len(all_covariates))]
        current_covariates = [all_covariates[j] for j, include in enumerate(selected_covariates_mask) if include]

        H, Y, teams = prepare_bt_covariate_data(df, year, current_covariates)
        model = sm.GLM(Y, H, family=sm.families.Binomial()).fit()

        # Retrieve AIC and BIC manually or via model
        aic = model.aic
        bic = model.bic

        results.append({
            "covariates": ", ".join(current_covariates) if current_covariates else "None",
            "aic": aic,
            "bic": bic,
            "model": model,
            "teams": teams
        })


    return pd.DataFrame(results)

# =========================================================
# RUN ACROSS ALL IPL YEARS
# =========================================================

all_years_bt_summary = []

for year in range(2008, 2020):
    print(f"Evaluating Bradley-Terry subsets for {year}...")

    results_df = evaluate_bt_subsets(merged_df_covariates, year)
    if results_df.empty:
        continue

    # Best subset by AIC
    best_aic_row = results_df.loc[results_df['aic'].idxmin()]
    # Best subset by BIC
    best_bic_row = results_df.loc[results_df['bic'].idxmin()]

    all_years_bt_summary.append({
        "year": year,
        "best_aic_covariates": best_aic_row['covariates'],
        "best_aic_val": best_aic_row['aic'],
        "best_bic_covariates": best_bic_row['covariates'],
        "best_bic_val": best_bic_row['bic']
    })

bt_best_subset_df = pd.DataFrame(all_years_bt_summary)
display(bt_best_subset_df)
```

```python
import pandas as pd
from collections import Counter
import numpy as np

bt_metrics_columns = {
    "BT_AIC": "best_aic_covariates",
    "BT_BIC": "best_bic_covariates"
}

all_bt_metric_frequency_tables = {}

for metric_name, col in bt_metrics_columns.items():
    cov_counter = Counter()

    for val in bt_best_subset_df[col]:
        if val is None or isinstance(val, float) and np.isnan(val):
            continue

        # cleaning
        val_str = str(val)
        val_str = val_str.replace("[", "").replace("]", "").replace("'", "").replace('"', "")
        val_str = val_str.strip()

        if val_str == "None" or val_str == "":
            continue

        covariates = [x.strip() for x in val_str.split(",") if x.strip() != ""]
        # Update counter
        cov_counter.update(covariates)

    # Create frequency table
    freq_df = pd.DataFrame(cov_counter.items(),columns=["Covariate", "Frequency"]).sort_values(by="Frequency", ascending=False).reset_index(drop=True)
    all_bt_metric_frequency_tables[metric_name] = freq_df
    print(f"\n{metric_name} INDIVIDUAL COVARIATE FREQUENCY")
    display(freq_df)
```
