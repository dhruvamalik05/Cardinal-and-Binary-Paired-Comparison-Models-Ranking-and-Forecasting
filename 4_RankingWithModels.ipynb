# **Ranking for Cardinal Model with 5 Covariates**

```python
import numpy as np
import itertools
import pandas as pd
from scipy.stats import kendalltau

selected_covariates = [
    'home_ground_advantage',
    'days_since_last_match_diff',
    'previous_match_advantage',
    'toss_winner_encoded',
    'rank_difference_prev_year'
]

def estimate_parameters(M, X, Y):
    if X is not None and X.shape[1] > 0:
        H = np.hstack([M, X])
    else:
        H = M
    theta_hat = np.linalg.lstsq(H, Y, rcond=None)[0]

    # K is the number of teams
    K_mu = M.shape[1]
    mu_hat = theta_hat[:K_mu]
    beta_hat = theta_hat[K_mu:]

    mu_hat = mu_hat - np.mean(mu_hat)

    return mu_hat, beta_hat, theta_hat

def prepare_ipl_data_for_year_with_covariates(merged_df, iplyear, covariates=None):
    ipl_year_df = merged_df[merged_df['year'] == iplyear].copy()

    unique_teams = sorted(list(pd.concat([ipl_year_df['team1'], ipl_year_df['team2']]).unique()))
    team_to_id = {team: i for i, team in enumerate(unique_teams)}
    K_teams = len(unique_teams)
    num_matches = len(ipl_year_df)

    covariates = list(covariates) if covariates is not None else []
    num_covariates = len(covariates)

    M = np.zeros((num_matches, K_teams))
    X_covariates = np.zeros((num_matches, num_covariates))
    Y = np.zeros(num_matches)

    for index, row in ipl_year_df.iterrows():
        match_array_idx = ipl_year_df.index.get_loc(index)

        team1_id = team_to_id[row['team1']]
        team2_id = team_to_id[row['team2']]

        M[match_array_idx, team1_id] = 1
        M[match_array_idx, team2_id] = -1

        if num_covariates > 0:
            X_covariates[match_array_idx, :] = row[covariates].values.astype(float)

        Y[match_array_idx] = row['win_by_runs']

    return M, X_covariates, Y, team_to_id, K_teams


final_result = {"year": [],
                "kendalltau":[],
                "pvalue": [],
                "residual_variance": [],
                "officialrank": [],
                "calculated rank": []}

for year in [2008, 2009, 2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019]:
  M, X, Y, team_to_id, K = prepare_ipl_data_for_year_with_covariates(merged_df_covariates, year, covariates = selected_covariates)
  print(X)

  rank_listed = {
    2008: [2, 7, 3, 1, 5, 4, 0, 6],
    2009: [1, 3, 0, 4, 7, 6, 5, 2],
    2010: [2, 1, 4, 7, 5, 0, 6, 3],
    2011: [1, 6, 9, 4, 7, 3, 2, 8, 5, 0],
    2012: [3, 7, 0, 5, 1, 2, 8, 6, 4],
    2013: [0, 8, 5, 6, 1, 7, 2, 4, 3],
    2014: [2, 7, 0, 1, 3, 4, 6, 5],
    2015: [0, 6, 7, 4, 1, 3, 2, 5],
    2016: [5, 0, 7, 3, 4, 6, 1, 2],
    2017: [5, 6, 4, 3, 0, 1, 7, 2],
    2018: [1, 7, 5, 2, 4, 3, 6, 0],
    2019: [1, 2, 5, 4, 0, 6, 7, 3]}

  res_var = np.nan

  mu_hat, beta_hat, theta_hat = estimate_parameters(M, X, Y)
  print("Team to ID mapping:", team_to_id)

  H = np.hstack([M, X])
  predictions = H @ theta_hat
  residuals = Y - predictions
  res_var = np.var(residuals)
  print(mu_hat)

  final_result["year"].append(year)
  final_result["kendalltau"].append(kendalltau(rank_listed[year], rank_scores(mu_hat))[0])
  final_result["pvalue"].append(kendalltau(rank_listed[year], rank_scores(mu_hat))[1])
  final_result["residual_variance"].append(res_var)
  final_result["officialrank"].append(rank_listed[year])
  final_result["calculated rank"].append(rank_scores(mu_hat))

pd.DataFrame(final_result)
```

# **Bradley Terry Implementation for Rankings**

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy.stats import kendalltau

bt_listed = {}

def prepare_bt_covariate_data(merged_df, iplyear, covariates):

    ipl_year_df = merged_df[merged_df['year'] == iplyear].copy()

    # Get unique teams for the year and create team_to_id mapping
    unique_teams = sorted(list(pd.concat([ipl_year_df['team1'], ipl_year_df['team2']]).unique()))
    team_to_id = {team: i for i, team in enumerate(unique_teams)}
    K_teams = len(unique_teams)

    num_matches = len(ipl_year_df)
    num_covariates = len(covariates)

    M = np.zeros((num_matches, K_teams))
    X_covariates = np.zeros((num_matches, num_covariates))
    Y = np.zeros(num_matches)

    for index, row in ipl_year_df.iterrows():
        match_array_idx = ipl_year_df.index.get_loc(index)

        team1_id = team_to_id[row['team1']]
        team2_id = team_to_id[row['team2']]

        # M matrix: 1 for team1, -1 for team2
        M[match_array_idx, team1_id] = 1
        M[match_array_idx, team2_id] = -1

        if num_covariates > 0:
            X_covariates[match_array_idx, :] = row[covariates].values.astype(float)

        Y[match_array_idx] = 1 if row['win_by_runs'] > 0 else 0

    if num_covariates > 0:
        H = np.hstack([M, X_covariates])
    else:
        H = M

    return H, Y, unique_teams, K_teams

all_bt_results = []

for year in range(2008, 2020):

    print(f"Running Bradley-Terry (GLM) for {year}...")
    if year not in rank_listed or len(rank_listed[year]) == 0:
        continue

    # PREPARE DESIGN MATRIX & OUTCOMES
    H, Y, teams,K = prepare_bt_covariate_data(merged_df_covariates,year,[])

    # FIT BRADLEY-TERRY MODEL VIA BINOMIAL GLM
    model = sm.GLM(Y,H,family=sm.families.Binomial())

    result = model.fit_regularized(alpha=1e-3,L1_wt=0)

    bt_strengths = result.params[:K]
    bt_strengths = bt_strengths - np.mean(bt_strengths)

    # COMPUTED RANKS
    calculated_rank = rank_scores(bt_strengths)
    bt_listed[year] = calculated_rank
    true_rank = rank_listed[year]


    # KENDALL TAU
    tau, p_value = kendalltau(true_rank,calculated_rank)

    # STORE RESULTS
    all_bt_results.append({
        "year": year,
        "kendall_tau bt no covariate": tau,
        "p_value": p_value,
        "true_rank": true_rank,
        "calculated_rank": calculated_rank,
        "strengths": bt_strengths.tolist()
    })

bt_results_df = pd.DataFrame(all_bt_results)
display(bt_results_df)
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy.stats import kendalltau

def compare_bt_models(df, covariates):
    results = []
    years = sorted(df['year'].unique())

    for year in years:


        # 1. BT WITHOUT COVARIATES
        H_no_cov, Y_no_cov, teams_no_cov, K_no_cov = prepare_bt_covariate_data(df, year, [])

        model_no_cov = sm.GLM(Y_no_cov, H_no_cov, family=sm.families.Binomial())
        result_no_cov = model_no_cov.fit_regularized(alpha=1e-4, L1_wt=0)

        bt_strengths = result_no_cov.params[:K_no_cov]
        bt_strengths = bt_strengths - np.mean(bt_strengths)
        rank_no_cov = rank_scores(bt_strengths)


        # 2. BT WITH COVARIATES (using Binomial GLM)
        H_cov, Y_cov, teams_cov,K_cov  = prepare_bt_covariate_data(df, year, covariates)

        model_cov = sm.GLM(Y_cov, H_cov, family=sm.families.Binomial())
        result_cov = model_cov.fit_regularized(alpha=1e-3, L1_wt=0)

        lambda_hat = result_cov.params[:K_cov]
        lambda_hat = lambda_hat - np.mean(lambda_hat)
        rank_with_cov = rank_scores(lambda_hat)


        # 3. COMPARE ALL MODELS
        tau_between, p_between = kendalltau(rank_no_cov, rank_with_cov)
        true_rank = rank_listed[year]
        tau_no_cov_vs_true, p_no_cov_vs_true = kendalltau(true_rank, rank_no_cov)
        tau_cov_vs_true, p_cov_vs_true = kendalltau(true_rank, rank_with_cov)

        results.append({
            'year': year,
            # 'teams': teams_no_cov,
            'true_rank': true_rank,
            'bt_no_covariate_rank': rank_no_cov,
            'bt_with_covariate_rank': rank_with_cov,
            'tau_no_cov_vs_true': tau_no_cov_vs_true,
            'p_no_cov_vs_true': p_no_cov_vs_true,
            'tau_cov_vs_true': tau_cov_vs_true,
            'p_cov_vs_true': p_cov_vs_true,
            'nocov_vs_cov_kendall_tau': tau_between,
            'nocov_vs_cov_p_value': p_between
        })

    return pd.DataFrame(results)

bt_comparison_summary_df = compare_bt_models(merged_df, selected_covariates)
display(bt_comparison_summary_df)
```
