```python
# def prepare_bt_covariate_data(
#     train_df,
#     year,
#     covariates
# ):

#     season_df = train_df[train_df['year'] == year].copy()
#     teams = sorted(list(pd.concat([season_df['team1'],season_df['team2']]).unique()))

#     team_to_id = {team: idx for idx, team in enumerate(teams)}

#     K = len(teams)
#     rows = []
#     outcomes = []

#     # -----------------------------------------------------
#     # BUILD DESIGN MATRIX
#     # -----------------------------------------------------

#     for _, row in season_df.iterrows():

#         team1 = row['team1']
#         team2 = row['team2']

#         i = team_to_id[team1]
#         j = team_to_id[team2]

#         # -------------------------------------------------
#         # BINARY OUTCOME
#         # -------------------------------------------------

#         y = 1 if row['win_by_runs'] > 0 else 0

#         # -------------------------------------------------
#         # TEAM DIFFERENCE VECTOR
#         # -------------------------------------------------

#         x_team = np.zeros(K)

#         x_team[i] = 1
#         x_team[j] = -1

#         # -------------------------------------------------
#         # COVARIATES
#         # -------------------------------------------------

#         x_cov = row[covariates].values.astype(float)

#         # -------------------------------------------------
#         # COMBINE
#         # -------------------------------------------------

#         full_row = np.concatenate([
#             x_team,
#             x_cov
#         ])

#         rows.append(full_row)
#         outcomes.append(y)

#     H = np.array(rows)
#     Y = np.array(outcomes)

#     return H, Y, teams


# =========================================================
# FIT BT MODEL
# =========================================================

def fit_bt_covariates(H, Y, K, len_covariates):

    model = sm.GLM(Y,H,family=sm.families.Binomial())
    # result = model.fit()

    team_penalties = np.full(K, 1e-3)
    covariate_penalties = np.full(len_covariates,1e-3)

    result = model.fit_regularized(alpha=np.concatenate([team_penalties, covariate_penalties]),L1_wt=0)

    return result


# =========================================================
# RANK EXTRACTION
# =========================================================

def extract_team_strengths(result, K):

    params = result.params

    # Team strengths only
    lambda_hat = params[:K]

    # Center strengths
    lambda_hat = lambda_hat - np.mean(lambda_hat)

    return lambda_hat


# =========================================================
# RANK SCORES
# =========================================================

def rank_scores(scores):
  rank = {i:scores[i] for i in range(len(scores))}
  rank = sorted(rank, key=lambda x: -rank[x])

  final_rank = [0]*len(rank)
  for i in range(len(rank)):
    rk = i
    team = rank[i]
    final_rank[team] = rk
  return final_rank



# =========================================================
# COVARIATES
# =========================================================

covariates = [

    'home_ground_advantage',
    'days_since_last_match_diff',
    'previous_match_advantage',
    'toss_winner_encoded',
    'rank_difference_prev_year'
]


import numpy as np
import pandas as pd
from scipy.linalg import pinv

def bt_walk_forward_prediction(merged_df_covariates, covariates, initial_fraction=0.5, k=0.5):
    results = []

    years = sorted(merged_df_covariates["year"].unique())

    for year in years:

        print(f"BT prediction {year}")
        season_df = merged_df_covariates[merged_df_covariates["year"] == year].reset_index(drop=True)
        n_matches = len(season_df)
        start_idx = int(initial_fraction * n_matches)

        correct = 0
        total = 0
        confidence_scores = []
        season_scores = []

        for t in range(start_idx, n_matches):
            train_df = season_df.iloc[:t]
            next_match = season_df.iloc[t]


            H, Y, teams, K = prepare_bt_covariate_data(train_df, year, covariates)

            # Fit Bradley-Terry model
            result = fit_bt_covariates(H, Y, K, len(covariates))
            strengths = extract_team_strengths(result, K)
            team_strength = {team: strengths[i] for i, team in enumerate(teams)}

            team1 = next_match["team1"]
            team2 = next_match["team2"]


            if team1 not in team_strength or team2 not in team_strength:
                continue

            # Calculate baseline paired strength difference
            score = team_strength[team1] - team_strength[team2]

            beta = result.params[K:]
            x_cov = next_match[covariates].astype(float).values
            score += np.dot(beta, x_cov)

            season_scores.append(score)

            # Statsmodels stores fitted probabilities in `result.mu` for GLM
            # p_l = result.mu
            linear_predictor = np.dot(H, result.params)
            p_l = 1 / (1 + np.exp(-linear_predictor))

            numerator = np.sum(((Y-p_l)**2)/(p_l*(1-p_l)))
            num_elements = len(Y)

            # --- 1. Compute H^T * W component using your exact conventions
            W_diag = p_l * (1 - p_l)
            alpha = 1e-3
            HtWH = (H.T * W_diag) @ H
            # --- 2. Compute Penalized Inverse Matrix with Uniform Alpha Penalty
            penalized_info_inv = np.linalg.inv(HtWH + alpha * np.eye(H.shape[1]))
            # --- 3. Compute Effective Parameters via Identity Trace Formula
            eff_parameters = np.trace(penalized_info_inv @ HtWH)

            nu = num_elements - eff_parameters

            scaled_numerator = (1/nu)*numerator

            denominator = np.sum(p_l*(1-p_l))
            scaled_denominator = (1/num_elements)*denominator

            Fisherweightedlogit_residstnddev = np.sqrt(scaled_numerator/scaled_denominator)

            threshold = Fisherweightedlogit_residstnddev

            if np.abs(score) < (k * threshold):
                continue

            # Formulate predictions vs actual labels
            pred = 1 if score > 0 else 0
            actual = 1 if next_match["win_by_runs"] > 0 else 0

            if pred == actual:
                correct += 1

            confidence_scores.append(abs(score))
            total += 1

        if len(season_scores) > 0:
            print(pd.Series(season_scores).describe())

        results.append({
            "year": year,
            "accuracy": correct / total if total > 0 else np.nan,
            "avg_confidence": np.mean(confidence_scores) if len(confidence_scores) else np.nan,
            "num_predictions": total
        })

    return pd.DataFrame(results)

# =====================================================
# CARDINAL WALK FORWARD
# =====================================================


# =====================================================
# CARDINAL DATA PREP
# =====================================================

# def prepare_ipl_data_for_year_with_covariates_df(
#     df,
#     covariates
# ):

#     teams = sorted(list(pd.concat([df["team1"],df["team2"]]).unique()))
#     team_to_id = {team: i for i, team in enumerate(teams)}
#     K = len(teams)

#     M_rows = []
#     X_rows = []
#     Y_rows = []

#     for _, row in df.iterrows():

#         i = team_to_id[row["team1"]]
#         j = team_to_id[row["team2"]]

#         m = np.zeros(K)
#         m[i] = 1
#         m[j] = -1
#         M_rows.append(m)

#         X_rows.append(row[covariates].astype(float).values)

#         Y_rows.append(row["win_by_runs"])

#     M = np.array(M_rows)
#     X = np.array(X_rows)
#     Y = np.array(Y_rows)

#     return (M,X,Y,team_to_id,K)


def cardinal_walk_forward_prediction(
    merged_df_covariates, covariates, initial_fraction=0.5, k=0.5):
    results = []
    years = sorted(merged_df_covariates["year"].unique())

    for year in years:
        print(f"Cardinal prediction {year}")
        season_df = (merged_df_covariates[merged_df_covariates["year"] == year].reset_index(drop=True))

        n_matches = len(season_df)
        start_idx = int(initial_fraction * n_matches)

        correct = 0
        total = 0
        confidence_scores = []
        season_scores = []

        for t in range(start_idx, n_matches):
            train_df = season_df.iloc[:t]
            next_match = season_df.iloc[t]

            # Prepare training data up to match t
            (M, X, Y, team_to_id, K) = prepare_ipl_data_for_year_with_covariates(
                train_df, year, covariates
            )
            H = np.hstack([M, X])


            # Fit the model parameters via pseudo-inverse
            theta = pinv(H.T @ H) @ H.T @ Y
            merits = theta[:K]

            team_names = sorted(team_to_id, key=team_to_id.get)
            merit_dict = {team_names[i]: merits[i] for i in range(K)}

            team1 = next_match["team1"]
            team2 = next_match["team2"]

            if team1 not in merit_dict or team2 not in merit_dict:
                continue

            score = merit_dict[team1] - merit_dict[team2]

            beta = theta[K:]
            x_cov = next_match[covariates].astype(float).values
            score += np.dot(beta, x_cov)

            season_scores.append(score)

            #  residual scale
            fitted_scores = Y - (H @ theta)
            threshold = np.std(fitted_scores)

            if np.abs(score) < k * threshold:
                continue

            # Predict 1 if team1 is favored, else 0
            pred = 1 if score > 0 else 0

            actual = 1 if next_match["win_by_runs"] > 0 else 0

            if pred == actual:
                correct += 1

            confidence_scores.append(abs(score))
            total += 1

        print(f"--- Season {year} Summary ---")
        print(pd.Series(season_scores).describe())

        results.append({
            "year": year,
            "accuracy": correct / total if total > 0 else np.nan,
            "avg_confidence": np.mean(confidence_scores) if len(confidence_scores) else np.nan,
            "num_predictions": total
        })

    return pd.DataFrame(results)

bt_home = bt_walk_forward_prediction(
    merged_df_covariates,
    ["home_ground_advantage"],
    k=0.75
)

bt_home_toss = bt_walk_forward_prediction(
    merged_df_covariates,
    [
        "home_ground_advantage",
        "toss_winner_encoded"
    ],
    k=0.75
)

card_home = cardinal_walk_forward_prediction(
    merged_df_covariates,
    ["home_ground_advantage"],
    k=0.75
)

card_home_toss = cardinal_walk_forward_prediction(
    merged_df_covariates,
    [
        "home_ground_advantage",
        "toss_winner_encoded"
    ],
    k=0.75
)

bt_home_rank_match = bt_walk_forward_prediction(
    merged_df_covariates,
    [
        "home_ground_advantage",
        "rank_difference_prev_year", "previous_match_advantage"
    ], k=0.75
)

card_home_rank_match = cardinal_walk_forward_prediction(
    merged_df_covariates,
    [
        "home_ground_advantage",
        "rank_difference_prev_year", "previous_match_advantage"
    ], k=0.75
)

# Helper to format accuracy with prediction counts
def fmt(acc, count):
    if pd.isna(acc):
        return "N/A"
    return f"{acc:.3f} ({count})"

comparison_table = pd.DataFrame({
    "year":
        bt_home["year"],

    "BT_Home":
        [fmt(a, c) for a, c in zip(bt_home["accuracy"], bt_home["num_predictions"])],

    "BT_Home_Toss":
        [fmt(a, c) for a, c in zip(bt_home_toss["accuracy"], bt_home_toss["num_predictions"])],

    "BT_Home_PrevRank_PrevMatch":
        [fmt(a, c) for a, c in zip(bt_home_rank_match["accuracy"], bt_home_rank_match["num_predictions"])],

    "Cardinal_Home":
        [fmt(a, c) for a, c in zip(card_home["accuracy"], card_home["num_predictions"])],

    "Cardinal_Home_Toss":
        [fmt(a, c) for a, c in zip(card_home_toss["accuracy"], card_home_toss["num_predictions"])],

    "Cardinal_Home_PrevRank_PrevMatch":
        [fmt(a, c) for a, c in zip(card_home_rank_match["accuracy"], card_home_rank_match["num_predictions"])]
})

display(comparison_table)
```
